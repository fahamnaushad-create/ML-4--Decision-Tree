##1. What is a Decision Tree, and how does it work in the context of classification?
- Answer:

    A decision tree is a flowchart-like model that splits data by asking a sequence of simple questions (feature thresholds) to separate classes. At each internal node the tree chooses a feature and a cutoff that best separates the samples; leaves assign class labels (usually by majority vote). For classification it recursively partitions the feature space so samples with similar target labels end up in the same leaf. Example: to classify flowers, the tree might first split on petal length > 2.5 cm, then on petal width, etc., until leaves predict species.

##2. Explain the concepts of Gini Impurity and Entropy as impurity measures. How do they impact the splits in a Decision Tree?
- Answer:

    Gini impurity and entropy measure how mixed the classes are in a node: lower values mean purer nodes. Gini impurity is 𝐺= 1 − ∑𝑝𝑖2 (fast to compute); entropy is 𝐻=−∑𝑝𝑖log⁡ 𝑝𝑖 (information-theoretic). When selecting a split the tree picks the feature/cut that yields the largest reduction in impurity (Gini decrease or information gain). Practically, Gini and entropy often pick similar splits; Gini is slightly faster and tends to emphasize larger class separations.

##3. What is the difference between Pre-Pruning and Post-Pruning in Decision Trees? Give one practical advantage of using each.
- Answer:

    Pre-pruning stops tree growth early using constraints (max_depth, min_samples_split) to prevent overfitting. Advantage: faster training and simpler model by design. Post-pruning grows a full tree first and then prunes back nodes (e.g., cost-complexity pruning) using validation data. Advantage: can find a better balance by removing branches that hurt generalization while preserving helpful splits discovered in the full tree.

##4. What is Information Gain in Decision Trees, and why is it important for choosing the best split?
- Answer:

    Information Gain is the reduction in impurity (usually entropy) after a split: IG = parent_entropy − weighted_child_entropy. It quantifies how much a split helps to make classes more pure. A split with higher information gain better separates classes and so is preferred when building the tree. Choosing splits by information gain leads to nodes that are more informative for prediction.

##5. What are some common real-world applications of Decision Trees, and what are their main advantages and limitations?
- Answer:

    Applications: medical diagnosis, credit scoring, customer churn prediction, fraud detection, and simple rule-based recommendation systems. Advantages: easy to interpret, handle mixed types, require little data preparation and reveal decision rules. Limitations: single trees can overfit, are unstable (small data changes can change splits), and usually underperform ensemble methods (Random Forest, XGBoost) on many complex problems.

##6. Write a Python program to:
    ● Load the Iris Dataset
    ● Train a Decision Tree Classifier using the Gini criterion
    ● Print the model’s accuracy and feature importances
    (Include your Python code and output.)
- Answer:

  ✅Python Code:

```python
# Q6: Iris + Decision Tree Classifier (Gini)
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np

iris = load_iris()
X, y = iris.data, iris.target

# 70% train, 30% test (stratify for balanced classes)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
feat_imp = clf.feature_importances_

print("Accuracy:", round(acc, 4))
print("Feature importances (sepal length, sepal width, petal length, petal width):")
print(np.round(feat_imp, 4))
```

##7. Write a Python program to:
###● Load the Iris Dataset

###● Train a Decision Tree Classifier with max_depth=3 and compare its accuracy to a fully-grown tree.
  (Include your Python code and output.)
  
- Answer:

  ✅Python Code:

```python
# Q7: compare max_depth=3 vs fully-grown tree on Iris
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

clf_depth3 = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)
clf_full = DecisionTreeClassifier(criterion='gini', random_state=42)  # no max_depth

clf_depth3.fit(X_train, y_train)
clf_full.fit(X_train, y_train)

acc_depth3 = accuracy_score(y_test, clf_depth3.predict(X_test))
acc_full = accuracy_score(y_test, clf_full.predict(X_test))

print("Accuracy (max_depth=3):", round(acc_depth3, 4))
print("Accuracy (fully-grown):", round(acc_full, 4))
```

##9. Write a Python program to:
● Load the Iris Dataset

● Tune the Decision Tree’s max_depth and min_samples_split using GridSearchCV

● Print the best parameters and the resulting model accuracy

(Include your Python code and output.)


  ✅Python Code:


```python
# Q9: GridSearchCV to tune max_depth and min_samples_split on Iris
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

param_grid = {
    'max_depth': [2, 3, 4, None],
    'min_samples_split': [2, 4, 6]
}

grid = GridSearchCV(DecisionTreeClassifier(criterion='gini', random_state=42),
                    param_grid, cv=5, scoring='accuracy', n_jobs=1)
grid.fit(X_train, y_train)

best_params = grid.best_params_
best_model = grid.best_estimator_

test_acc = accuracy_score(y_test, best_model.predict(X_test))

print("Best parameters:", best_params)
print("Test set accuracy with best parameters:", round(test_acc, 4))
```

    - Step-by-step process (clear, practical):

    3. Encode categorical features: For low-cardinality categories use one-hot encoding; for higher cardinality use target encoding or ordinal encoding if there is order. For Decision Trees, ordinal/label encoding can work because tree splits on values; still be careful with label encoding of nominal categories — one-hot is safer for non-ordinal data.

    4. Feature engineering & scaling: Trees don’t need standard scaling, but create informative features (e.g., BMI from height/weight, grouped age bins). Convert timestamps to meaningful features. Remove or treat highly correlated redundant features if they hurt interpretability.

    7. Evaluation & calibration: Evaluate with appropriate metrics—AUC-ROC, precision, recall, F1-score, and especially precision-recall curves if disease is rare. Use confusion matrix to inspect false positives/negatives. Consider calibrating probabilities with isotonic or Platt scaling (CalibratedClassifierCV) if probability estimates are needed for decisions.

    8. Interpretability & validation: Use feature importances and SHAP/partial dependence plots to explain model decisions. Validate findings with clinicians to ensure medically plausible behavior. Check fairness and potential biases across subgroups.

    9. Business value: A reliable disease-prediction model can flag high-risk patients for early screening, prioritize clinical tests, reduce time-to-diagnosis, and allocate resources efficiently. It improves patient outcomes by enabling preventive care and can lower costs by avoiding unnecessary broad testing—provided model predictions are validated clinically and used with human oversight.
